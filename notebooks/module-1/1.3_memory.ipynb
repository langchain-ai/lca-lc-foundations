```python
from dotenv import load_dotenv
import os
from pydantic import BaseModel
from langgraph.checkpoint.memory import InMemorySaver
from langchain.messages import HumanMessage
from langchain.agents import create_agent
from pprint import pprint

load_dotenv()
model=os.getenv("OPENAI_MODEL_NAME")
```

## No memory

```python
agent = create_agent(
    model=model
)
question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")

response = agent.invoke(
    {"messages": [question]} 
)
pprint(response)
pprint(response['messages'][-1].content)
```

```python
question = HumanMessage(content="What's my favourite colour?")
response = agent.invoke(
    {"messages": [question]} 
)
pprint(response)
```

## Memory

```python
system_prompt="""
You are a concise question-answering assistant. 
Answer only what the user asks using the provided context. 
Keep answers as short as possible, using minimum tokens. 
Do not add explanations, extra information, or follow-up questions. 
If the answer is not in the context, respond: I don't know."""

agent = create_agent(
    model=model,
    checkpointer=InMemorySaver(),  
)
```

```python
question = HumanMessage(content="Hello my name is Seán and my favourite colour is green and my age is 22 lives in india. Studying Engineering")
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config,  
)
pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")
response = agent.invoke(
    {"messages": [question]},
    config,  
)
pprint(response)
pprint(response['messages'][-1].content)
```

```python
question = HumanMessage(content="What's my name")

response = agent.invoke(
    {"messages": [question]},
    config,  
)

pprint(response)
pprint(response['messages'][-1].content)
```
